## Exercise 1: Messy Sensor Data Cleaner
**Objective:** Practice type conversion and error handling (`try/except`).

You are given a list of temperature readings from a faulty sensor. Some are integers, some are strings, and some are completely invalid data (like `None` or text). 

Write a function that loops through the list, attempts to convert each item to a float, and ignores invalid entries using a `try/except` block. Finally, return the average of the valid temperatures.

````python
def clean_and_average(raw_data):
    valid_temps = []

    # TODO: Loop through raw_data
    for elem in raw_data:

    # TODO: Use try/except to convert each item to a float
        try:
            temp = float(elem)

    # TODO: Append valid floats to valid_temps
            valid_temps.append(temp)
        except ValueError:
            print(f"Skipping {elem}: cannot convert value to float")
        except TypeError:
            print(f"Skipping {elem}: wrong data type")

    # TODO: Calculate and return the average
    total = 0
    for elem in valid_temps:
        total = total + elem
    avg_div = total / len(valid_temps)
    return avg_div

sensor_readings = [22.5, "23.1", "error", None, 21, "22.0C", 19.8]
print(f"Average: {clean_and_average(sensor_readings)}")
````

## Exercise 5: Universal Unit Converter
**Objective:** Build a menu-driven unit converter with robust exception handling and clean function design.

Create a program that converts between different unit systems. The user selects a category from a menu, then inputs a value to convert.

**Requirements:**
- Support at least 3 categories: **Temperature** (°C ↔ °F ↔ K), **Distance** (km ↔ miles ↔ meters), and **Weight** (kg ↔ lbs ↔ grams).
- Each category must be handled by its own dedicated function.
- Use `try/except` to catch `ValueError` when the user enters non-numeric input.
- Use a `while` loop so the user can perform multiple conversions before choosing to quit.
- **Edge Case:** Reject physically impossible values (e.g., a temperature below absolute zero: -273.15°C).

````python
def convert_temperature(value, from_unit, to_unit):
    """Convert between Celsius, Fahrenheit, and Kelvin."""

    # TODO: Raise ValueError if result is below absolute zero
    if from_unit == "c" and value < -273.15:
        raise ValueError("Temperature cannot be below absolute zero.")
    
    elif from_unit == "k" and value < 0:
        raise ValueError("Temperature cannot be below absolute zero.")
    
    elif from_unit == "f" and value < -459.67:
        raise ValueError("Temperature cannot be below absolute zero.")

    # TODO: Implement conversion logic
    if from_unit == "c" and to_unit == "f":
        f = value * 9/5 + 32
        return f

    elif from_unit == "f" and to_unit == "c":
        c = (value - 32) * 5/9
        return c

    elif from_unit == "c" and to_unit == "k":
        k = value + 273.15
        return k

    elif from_unit == "k" and to_unit == "c":
        c = value - 273.15
        return c

    elif from_unit == "f" and to_unit == "k":
        k = (value - 32) * 5/9 + 273.15
        return k

    elif from_unit == "k" and to_unit == "f":
        f = (value - 273.15) * 9/5 + 32
        return f

    elif from_unit == to_unit:
        return value

    

def convert_distance(value, from_unit, to_unit):
    """Convert between km, miles, and meters."""

    # TODO: Reject negative distances
    if value < 0:
        raise ValueError("Distance cannot be negative")

    # TODO: Implement conversion logic
    if from_unit == "km" and to_unit == "miles":
        miles = value * 0.621371
        return miles

    elif from_unit == "miles" and to_unit == "km":
        km = value / 0.621371
        return km

    elif from_unit == "km" and to_unit == "meters":
        meters = value * 1000
        return meters

    elif from_unit == "meters" and to_unit == "km":
        km = value / 1000
        return km

    elif from_unit == "miles" and to_unit == "meters":
        meters = value * 1609.344
        return meters

    elif from_unit == "meters" and to_unit == "miles":
        miles = value / 1609.344
        return miles

    elif from_unit == to_unit:
        return value

def convert_weight(value, from_unit, to_unit):
    """Convert between kg, lbs, and grams."""

    # TODO: Reject negative weights
    if value < 0:
        raise ValueError("Weight cannot be negative.")

    # TODO: Implement conversion logic
    if from_unit == "kg" and to_unit == "lbs":
        lbs = value * 2.20462
        return lbs

    elif from_unit == "lbs" and to_unit == "kg":
        kg = value / 2.20462 
        return kg

    elif from_unit == "kg" and to_unit == "grams":
        grams = value * 1000 
        return grams

    elif from_unit == "grams" and to_unit == "kg":
        kg = value / 1000
        return kg

    elif from_unit == "lbs" and to_unit == "grams":
        grams = value * 453.592
        return grams

    elif from_unit == "grams" and to_unit == "lbs":
        lbs = value / 453.592
        return lbs

    elif from_unit == to_unit:
        return value

def run_converter():
    """Main loop: display menu, get user choice, call the right function."""

    # TODO: while True loop with menu: [1] Temperature [2] Distance [3] Weight [Q] Quit
    while True:
        print("\nUnit Converter")
        print("[1] Temperature")
        print("[2] Distance")
        print("[3] Weight")
        print("[Q] Quit")

        choice = input("Choose an option: ").strip().lower()

        if choice == "q":
            print("Goodbye!")
            break

        try:

    # TODO: Inside each option, ask for value, from_unit, to_unit
            if choice == "1":
                value = float(input("Enter temperature: "))
                from_unit = input("Convert from (c/f/k): ").strip().lower()
                to_unit = input("Convert to (c/f/k): ").strip().lower()

                if from_unit not in ["c", "f", "k"] or to_unit not in ["c", "f", "k"]:
                    raise ValueError("Invalid temperature unit.")

                result = convert_temperature(value, from_unit, to_unit)
                print(f"Result: {result:.2f} {to_unit.capitalize()}")

            elif choice == "2":
                value = float(input("Enter distance: "))
                from_unit = input("Convert from (km/miles/meters): ").strip().lower()
                to_unit = input("Convert to (km/miles/meters): ").strip().lower()

                if from_unit not in ["km", "miles", "meters"] or to_unit not in ["km", "miles", "meters"]:
                    raise ValueError("Invalid distance unit.")
        
                result = convert_distance(value, from_unit, to_unit)                
                print(f"Result: {result:.2f} {to_unit}")

            elif choice == "3":
                value = float(input("Enter weight: "))
                from_unit = input("Convert from (kg/grams/lbs): ").strip().lower()
                to_unit = input("Convert to (kg/grams/lbs): ").strip().lower()

                if from_unit not in ["kg", "grams", "lbs"] or to_unit not in ["kg", "grams", "lbs"]:
                    raise ValueError("Invalid weight unit.")

                result = convert_weight(value, from_unit, to_unit)                
                print(f"Result: {result:.2f} {to_unit}")

    # TODO: Wrap everything in try/except to catch ValueError and bad input
            else: 
                print("Invalid option. Please choose 1, 2, 3 or Q.")

        except ValueError as error:
            print(f"Error: {error}")

run_converter()  # Uncomment to test
````
