## Exercise 1: Messy Sensor Data Cleaner
**Objective:** Practice type conversion and error handling (`try/except`).

You are given a list of temperature readings from a faulty sensor. Some are integers, some are strings, and some are completely invalid data (like `None` or text). 

Write a function that loops through the list, attempts to convert each item to a float, and ignores invalid entries using a `try/except` block. Finally, return the average of the valid temperatures.

````python
def clean_and_average(raw_data):
    valid_temps = []
    for data in raw_data:
        print("Converting:", data, " to float")
        try:
            temp = float(data)
            valid_temps.append(temp)
        except:
            print("Error, item cannot be converted")
    sum = 0
    for i in valid_temps:
        sum += i
    return sum / len(valid_temps)


sensor_readings = [22.5, "23.1", "error", None, 21, "22.0C", 19.8]
print(f"Average: {clean_and_average(sensor_readings):.2f}")
````

## Exercise 2: Transit Kiosk Menu Simulator
**Objective:** Build a menu-driven application with nested conditionals and defensive input validation.

Create a `while` loop that displays a menu for a transit ticket kiosk. Ask the user for their destination zone (1, 2, or 3) and their age. Apply a 50% discount if they are under 18 or over 65. 

Use `try/except` to ensure they don't crash the program by typing letters instead of numbers. If they enter invalid input, print an error and restart the loop.

````python
def run_kiosk():
    while True:
        try:
            zone = int(input("Zone (1-3): "))
            if zone not in (1, 2, 3):
                print("Error: please choose zone 1, 2, or 3.")
                continue

            age = int(input("Age: "))
            if age < 0:
                print("Error: age cannot be negative.")
                continue
        except ValueError:
            print("Error: zone and age must be numbers.")
            continue

        if age < 18 or age > 65:
            if zone == 1:
                price = 2.50 * 0.50
            elif zone == 2:
                price = 4.00 * 0.50
            else:
                price = 6.50 * 0.50
        else:
            if zone == 1:
                price = 2.50
            elif zone == 2:
                price = 4.00
            else:
                price = 6.50

        print(f"Ticket for Zone {zone}: ${price:.2f}")

        again = input("Buy another ticket? (yes/no): " ).strip().lower()
        if again not in ("yes"):
            break

run_kiosk() # Uncomment to test
````

## Exercise 5: Universal Unit Converter
**Objective:** Build a menu-driven unit converter with robust exception handling and clean function design.

Create a program that converts between different unit systems. The user selects a category from a menu, then inputs a value to convert.

**Requirements:**
- Support at least 3 categories: **Temperature** (°C ↔ °F ↔ K), **Distance** (km ↔ miles ↔ meters), and **Weight** (kg ↔ lbs ↔ grams).
- Each category must be handled by its own dedicated function.
- Use `try/except` to catch `ValueError` when the user enters non-numeric input.
- Use a `while` loop so the user can perform multiple conversions before choosing to quit.
- **Edge Case:** Reject physically impossible values (e.g., a temperature below absolute zero: -273.15°C).

````python
def convert_temperature(value, from_unit, to_unit):
    units = {"c": "C", "°c": "C", "celsius": "C", "f": "F", "°f": "F", "fahrenheit": "F", "k": "K", "kelvin": "K"}
    try:
        source, target = units[from_unit.strip().lower()], units[to_unit.strip().lower()]
    except KeyError as error:
        raise ValueError("Use C, F, or K for temperature units.") from error
    celsius = value if source == "C" else (value - 32) * 5 / 9 if source == "F" else value - 273.15
    if celsius < -273.15:
        raise ValueError("Temperature cannot be below absolute zero.")
    return celsius if target == "C" else celsius * 9 / 5 + 32 if target == "F" else celsius + 273.15


def convert_distance(value, from_unit, to_unit):
    units = {"km": 1000, "kilometer": 1000, "kilometers": 1000, "mi": 1609.344, "mile": 1609.344, "miles": 1609.344, "m": 1, "meter": 1, "meters": 1}
    try:
        source, target = units[from_unit.strip().lower()], units[to_unit.strip().lower()]
    except KeyError as error:
        raise ValueError("Use km, miles, or meters for distance units.") from error
    if value < 0:
        raise ValueError("Distance cannot be negative.")
    return value * source / target


def convert_weight(value, from_unit, to_unit):
    units = {"kg": 1000, "kilogram": 1000, "kilograms": 1000, "lb": 453.59237, "lbs": 453.59237, "pound": 453.59237, "pounds": 453.59237, "g": 1, "gram": 1, "grams": 1}
    try:
        source, target = units[from_unit.strip().lower()], units[to_unit.strip().lower()]
    except KeyError as error:
        raise ValueError("Use kg, lbs, or grams for weight units.") from error
    if value < 0:
        raise ValueError("Weight cannot be negative.")
    return value * source / target


def run_converter():
    converters = {"1": ("Temperature", convert_temperature), "2": ("Distance", convert_distance), "3": ("Weight", convert_weight)}
    while True:
        print("Categories: [1] Temperature  [2] Distance  [3] Weight  [Q] Quit")
        choice = input("Choose a category: ").strip().lower()
        if choice == "q":
            print("Goodbye!")
            break
        if choice not in converters:
            print("Invalid choice. Select 1, 2, 3, or Q.")
            continue
        category, converter = converters[choice]
        try:
            value = float(input("Value to convert: "))
            from_unit = input("Convert from unit: ")
            to_unit = input("Convert to unit: ")
            result = converter(value, from_unit, to_unit)
            print(f"{value:g} {from_unit.strip()} = {result:.4f} {to_unit.strip()}")
        except ValueError as error:
            print(f"{category} conversion error: {error}")

run_converter()  # Uncomment to test
````

## Exercise 6: Password Strength Validator
**Objective:** Practice string methods, conditional logic, and loop-based analysis.

Write a function that evaluates the strength of a password and returns a score from 0 to 5. The user can keep testing passwords until they type `quit`.

**Scoring Rules (1 point each):**
1. Length is at least 8 characters.
2. Contains at least one uppercase letter.
3. Contains at least one lowercase letter.
4. Contains at least one digit.
5. Contains at least one special character (`!@#$%^&*`).

**Output:**
- 0–1: `Weak`
- 2–3: `Medium`
- 4–5: `Strong`

**Hint:** Use a `for` loop to iterate through each character of the password string and check its properties with `.isupper()`, `.isdigit()`, etc.

````python
def check_password_strength(password):
    score = 0
    special_chars = "!@#$%^&*"
    has_upper = False
    has_lower = False
    has_digit = False
    has_special = False

    if len(password) >= 8:
        score += 1

    for character in password:
        if character.isupper():
            has_upper = True
        elif character.islower():
            has_lower = True
        elif character.isdigit():
            has_digit = True
        elif character in special_chars:
            has_special = True

    score += sum((has_upper, has_lower, has_digit, has_special))
    if score <= 1:
        label = "Weak"
    elif score <= 3:
        label = "Medium"
    else:
        label = "Strong"
    return score, label


def run_password_checker():
    while True:
        password = input("Enter a password (or 'quit' to exit): ")
        if password.strip().lower() == "quit":
            print("Goodbye!")
            break
        score, label = check_password_strength(password)
        print(f"Strength: {label} ({score}/5)")

run_password_checker()  # Uncomment to test
````
