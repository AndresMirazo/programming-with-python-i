````python
def convert_temperature(value, from_unit, to_unit):
    """Convert between Celsius, Fahrenheit, and Kelvin."""
    from_unit = from_unit.strip().upper()
    to_unit = to_unit.strip().upper()

    if from_unit not in ("C", "F", "K") or to_unit not in ("C", "F", "K"):
        raise ValueError("Invalid unit. Use C, F, or K.")

    if from_unit == "C":
        celsius = value
    elif from_unit == "F":
        celsius = (value - 32) * 5 / 9
    else:
        celsius = value - 273.15

    if celsius < -273.15:
        raise ValueError("Temperature is below absolute zero (-273.15 °C).")

    if to_unit == "C":
        return celsius
    elif to_unit == "F":
        return celsius * 9 / 5 + 32
    else:
        return celsius + 273.15


def convert_distance(value, from_unit, to_unit):
    """Convert between km, miles, and meters."""
    if value < 0:
        raise ValueError("Distance cannot be negative.")

    to_meters = {"km": 1000, "miles": 1609.344, "m": 1}

    from_unit = from_unit.strip().lower()
    to_unit = to_unit.strip().lower()

    if from_unit not in to_meters or to_unit not in to_meters:
        raise ValueError("Invalid unit. Use km, miles, or m.")

    return value * to_meters[from_unit] / to_meters[to_unit]


def convert_weight(value, from_unit, to_unit):
    """Convert between kg, lbs, and grams."""
    if value < 0:
        raise ValueError("Weight cannot be negative.")

    to_kg = {"kg": 1, "lbs": 0.45359237, "g": 0.001}

    from_unit = from_unit.strip().lower()
    to_unit = to_unit.strip().lower()

    if from_unit not in to_kg or to_unit not in to_kg:
        raise ValueError("Invalid unit. Use kg, lbs, or g.")

    return value * to_kg[from_unit] / to_kg[to_unit]


def run_converter():
    """Main loop: display menu, get user choice, call the right function."""
    while True:
        print("\n=== Unit Converter ===")
        print("[1] Temperature  [2] Distance  [3] Weight  [Q] Quit")
        choice = input("Choose an option: ").strip().lower()

        if choice == "q":
            print("Goodbye!")
            break

        if choice not in ("1", "2", "3"):
            print("Invalid option. Choose 1, 2, 3, or Q.")
            continue

        try:
            value = float(input("Value to convert: "))

            if choice == "1":
                from_unit = input("From (C, F, K): ")
                to_unit = input("To (C, F, K): ")
                result = convert_temperature(value, from_unit, to_unit)
            elif choice == "2":
                from_unit = input("From (km, miles, m): ")
                to_unit = input("To (km, miles, m): ")
                result = convert_distance(value, from_unit, to_unit)
            else:
                from_unit = input("From (kg, lbs, g): ")
                to_unit = input("To (kg, lbs, g): ")
                result = convert_weight(value, from_unit, to_unit)

            print(f"Result: {value} {from_unit} = {result:.4f} {to_unit}")

        except ValueError as e:
            print(f"Error: {e}")


run_converter()  # Uncomment to test
````
