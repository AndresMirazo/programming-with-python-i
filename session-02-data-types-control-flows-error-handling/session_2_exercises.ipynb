## Exercise 1: Messy Sensor Data Cleaner
**Objective:** Practice type conversion and error handling (`try/except`).

You are given a list of temperature readings from a faulty sensor. Some are integers, some are strings, and some are completely invalid data (like `None` or text). 

Write a function that loops through the list, attempts to convert each item to a float, and ignores invalid entries using a `try/except` block. Finally, return the average of the valid temperatures.

````python
def clean_and_average(raw_data):

    valid_temps = []

    for item in raw_data:
        try:
            temp = float(item)
            valid_temps.append(temp)
        except (ValueError, TypeError):
            pass

    return sum(valid_temps) / len(valid_temps)


sensor_readings = [22.5, "23.1", "error", None, 21, "22.0C", 19.8]

print(f"Average: {clean_and_average(sensor_readings)}")
````

Use `try/except` to ensure they don't crash the program by typing letters instead of numbers. If they enter invalid input, print an error and restart the loop.

## Exercise 3: Three-Strike Quiz System
**Objective:** Use loops, counters, and conditionals to manage state.

Write a short quiz program. The user must answer a question correctly. They have exactly 3 attempts.

- If they fail 3 times, print a failure message.
- **Defensive Code:** Ensure you handle edge cases like different capitalizations or extra spaces in their input (e.g., " Paris " should be accepted).

````python
def run_quiz():
    question = "What is the capital of France?"
    correct_answer = "paris"
    attempts = 3


    while attempts > 0:
        user_answer = input(f"{question} (Attempts left: {attempts}): ")

        cleaned_answer = user_answer.strip().lower()

        if cleaned_answer == correct_answer:
            print("Correct!")
            break
        else:
            attempts -= 1
            print("Incorrect.")

    if attempts == 0:
        print("No attempts left. The correct answer is Paris.")


run_quiz()
````

## Exercise 5: Universal Unit Converter
**Objective:** Build a menu-driven unit converter with robust exception handling and clean function design.

Create a program that converts between different unit systems. The user selects a category from a menu, then inputs a value to convert.

**Requirements:**
- Support at least 3 categories: **Temperature** (°C ↔ °F ↔ K), **Distance** (km ↔ miles ↔ meters), and **Weight** (kg ↔ lbs ↔ grams).
- Each category must be handled by its own dedicated function.
- Use `try/except` to catch `ValueError` when the user enters non-numeric input.
- Use a `while` loop so the user can perform multiple conversions before choosing to quit.
- **Edge Case:** Reject physically impossible values (e.g., a temperature below absolute zero: -273.15°C).

````python
def convert_temperature(value, from_unit, to_unit):
    try:
        if from_unit == "c" and to_unit == "f":
            result = value * 9 / 5 + 32

        elif from_unit == "f" and to_unit == "c":
            result = (value - 32) * 5 / 9

        elif from_unit == "c" and to_unit == "k":
            result = value + 273.15

        elif from_unit == "k" and to_unit == "c":
            result = value - 273.15

        elif from_unit == "f" and to_unit == "k":
            result = (value - 32) * 5 / 9 + 273.15

        elif from_unit == "k" and to_unit == "f":
            result = (value - 273.15) * 9 / 5 + 32

        elif from_unit == to_unit:
            result = value

        else:
            raise ValueError("Invalid temperature unit")

        if result < -273.15 and to_unit == "c":
            raise ValueError("Temperature is below absolute zero")

        if result < 0 and to_unit == "k":
            raise ValueError("Temperature is below absolute zero")

        if result < -459.67 and to_unit == "f":
            raise ValueError("Temperature is below absolute zero")

        return result

    except ValueError as e:
        print("Error:", e)
        return None


def convert_distance(value, from_unit, to_unit):
    try:
        if value < 0:
            raise ValueError("Distance cannot be negative")

        if from_unit == "km" and to_unit == "m":
            return value * 1000

        elif from_unit == "m" and to_unit == "km":
            return value / 1000

        elif from_unit == "km" and to_unit == "mi":
            return value * 0.621371

        elif from_unit == "mi" and to_unit == "km":
            return value / 0.621371

        elif from_unit == "m" and to_unit == "mi":
            return value / 1609.344

        elif from_unit == "mi" and to_unit == "m":
            return value * 1609.344

        elif from_unit == to_unit:
            return value

        else:
            raise ValueError("Invalid distance unit")

    except ValueError as e:
        print("Error:", e)
        return None


def convert_weight(value, from_unit, to_unit):
    try:
        if value < 0:
            raise ValueError("Weight cannot be negative")

        if from_unit == "kg" and to_unit == "g":
            return value * 1000

        elif from_unit == "g" and to_unit == "kg":
            return value / 1000

        elif from_unit == "kg" and to_unit == "lb":
            return value * 2.20462

        elif from_unit == "lb" and to_unit == "kg":
            return value / 2.20462

        elif from_unit == "g" and to_unit == "lb":
            return value / 453.592

        elif from_unit == "lb" and to_unit == "g":
            return value * 453.592

        elif from_unit == to_unit:
            return value

        else:
            raise ValueError("Invalid weight unit")

    except ValueError as e:
        print("Error:", e)
        return None


def run_converter():

    while True:
        print("\n1. Temperature")
        print("2. Distance")
        print("3. Weight")
        print("Q. Quit")

        choice = input("Choose an option: ").lower()

        if choice == "q":
            print("Goodbye!")
            break

        try:
            value = float(input("Enter value: "))
            from_unit = input("From unit: ").lower()
            to_unit = input("To unit: ").lower()

            if choice == "1":
                result = convert_temperature(value, from_unit, to_unit)

            elif choice == "2":
                result = convert_distance(value, from_unit, to_unit)

            elif choice == "3":
                result = convert_weight(value, from_unit, to_unit)

            else:
                print("Invalid option")
                continue

            if result is not None:
                print("Result:", result)

        except ValueError:
            print("Please enter a number")


run_converter()
````
