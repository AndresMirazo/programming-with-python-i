## Exercise 1: Messy Sensor Data Cleaner
**Objective:** Practice type conversion and error handling (`try/except`).

You are given a list of temperature readings from a faulty sensor. Some are integers, some are strings, and some are completely invalid data (like `None` or text). 

Write a function that loops through the list, attempts to convert each item to a float, and ignores invalid entries using a `try/except` block. Finally, return the average of the valid temperatures.

````python
def clean_and_average(raw_data):
    valid_temps = []
    for elem in raw_data:
        try:
            valid_temps.append(float(elem))
        except(TypeError, ValueError):
            pass
    avg=0
    totalsum=0
    for elem in valid_temps:
        totalsum= totalsum + elem 
        avg = totalsum/len(valid_temps)
    return avg

sensor_readings = [22.5, "23.1", "error", None, 21, "22.0C", 19.8]
print(f"Average: {clean_and_average(sensor_readings)}")
````

## Exercise 5: Universal Unit Converter
**Objective:** Build a menu-driven unit converter with robust exception handling and clean function design.

Create a program that converts between different unit systems. The user selects a category from a menu, then inputs a value to convert.

**Requirements:**
- Support at least 3 categories: **Temperature** (°C ↔ °F ↔ K), **Distance** (km ↔ miles ↔ meters), and **Weight** (kg ↔ lbs ↔ grams).
- Each category must be handled by its own dedicated function.
- Use `try/except` to catch `ValueError` when the user enters non-numeric input.
- Use a `while` loop so the user can perform multiple conversions before choosing to quit.
- **Edge Case:** Reject physically impossible values (e.g., a temperature below absolute zero: -273.15°C).

````python
def convert_temperature(value, from_unit, to_unit):
    """Convert between Celsius, Fahrenheit, and Kelvin."""
    # TODO: Implement conversion logic
    # TODO: Raise ValueError if result is below absolute zero
    try:
        value = float(value)
    except ValueError:
        raise ValueError("The temperature value must be a number.")

    from_u = from_unit.strip().capitalize()
    to_u = to_unit.strip().capitalize()

    valid_units = ["Celsius", "Fahrenheit", "Kelvin"]

    if from_u not in valid_units:
        raise ValueError(f"Unknown unit: '{from_unit}'. Must be Celsius, Fahrenheit, or Kelvin.")

    if to_u not in valid_units:
        raise ValueError(f"Unknown unit: '{to_unit}'. Must be Celsius, Fahrenheit, or Kelvin.")

    if from_u == to_u:
        raise ValueError(f"Cannot convert from {from_unit} to the same unit ({to_unit}).")

    if from_u == "Celsius" and value < -273.15:
        raise ValueError("Temperature below absolute zero (-273.15°C) is physically impossible.")
    elif from_u == "Fahrenheit" and value < -459.67:
        raise ValueError("Temperature below absolute zero (-459.67°F) is physically impossible.")
    elif from_u == "Kelvin" and value < 0:
        raise ValueError("Temperature below absolute zero (0 K) is physically impossible.")

    if from_u == "Celsius" and to_u == "Fahrenheit":
        return (value * 9 / 5) + 32
    
    elif from_u == "Celsius" and to_u == "Kelvin":
        return value + 273.15

    elif from_u == "Fahrenheit" and to_u == "Celsius":
        return (value - 32) * 5 / 9

    elif from_u == "Fahrenheit" and to_u == "Kelvin":
        return (value - 32) * 5 / 9 + 273.15
    
    elif from_u == "Kelvin" and to_u == "Celsius":
        return value - 273.15

    elif from_u == "Kelvin" and to_u == "Fahrenheit":
        return (value - 273.15) * 9 / 5 + 32

def convert_distance(value, from_unit, to_unit):
    """Convert between Km, Miles, and Meters."""

    try:
        value = float(value)
    except ValueError:
        raise ValueError("The distance value must be a number.")
    
    if value < 0:
        raise ValueError("Distance cannot be negative.")

    from_u = from_unit.strip().capitalize()
    to_u = to_unit.strip().capitalize()
    
    valid_units = ["Km", "Miles", "Meters"]

    if from_u not in valid_units:
        raise ValueError(f"Unknown unit: '{from_unit}'. Must be Km, Miles, or Meters.")

    if to_u not in valid_units:
        raise ValueError(f"Unknown unit: '{to_unit}'. Must be Km, Miles, or Meters.")

    if from_u == to_u:
        raise ValueError(f"Cannot convert from {from_unit} to the same unit ({to_unit}).")

    if from_u == "Km" and to_u == "Miles":
        return value * 0.621371

    elif from_u == "Km" and to_u == "Meters":
        return value * 1000

    elif from_u == "Miles" and to_u == "Km":
        return value * 1.60934

    elif from_u == "Miles" and to_u == "Meters":
        return value * 1609.34

    elif from_u == "Meters" and to_u == "Km":
        return value / 1000

    elif from_u == "Meters" and to_u == "Miles":
        return value / 1609.34

def convert_weight(value, from_unit, to_unit):
    """Convert between Kg, Lbs, and Grams."""

    try:
        value = float(value)
    except ValueError:
        raise ValueError("The weight value must be a number.")
    
    if value < 0:
        raise ValueError("Weight cannot be negative.")

    from_u = from_unit.strip().capitalize()
    to_u = to_unit.strip().capitalize()
    
    valid_units = ["Kg", "Lbs", "Grams"]

    if from_u not in valid_units:
        raise ValueError(f"Unknown unit: '{from_unit}'. Must be Kg, Lbs, or Grams.")

    if to_u not in valid_units:
        raise ValueError(f"Unknown unit: '{to_unit}'. Must be Kg, Lbs, or Grams.")

    if from_u == to_u:
        raise ValueError(f"Cannot convert from {from_unit} to the same unit ({to_unit}).")

    if from_u == "Kg" and to_u == "Lbs":
        return value * 2.20462

    elif from_u == "Kg" and to_u == "Grams":
        return value * 1000

    elif from_u == "Lbs" and to_u == "Kg":
        return value * 0.453592

    elif from_u == "Lbs" and to_u == "Grams":
        return value * 453.592

    elif from_u == "Grams" and to_u == "Kg":
        return value / 1000

    elif from_u == "Grams" and to_u == "Lbs":
        return value / 453.592

def run_converter():
    """Main loop: display menu, get user choice, call the right function."""
    # TODO: while True loop with menu: [1] Temperature [2] Distance [3] Weight [Q] Quit
    # TODO: Inside each option, ask for value, from_unit, to_unit
    # TODO: Wrap everything in try/except to catch ValueError and bad input
    
    while True:
        answer_main = input("\nChoose a conversion: [1] Temperature [2] Distance [3] Weight [Q] Quit: ").strip().upper()

        if answer_main == "Q":
            print("Goodbye!")
            break

        elif answer_main == "1":
            print("\n--- Temperature Converter ---")
            print("Available units: [1] Celsius  [2] Fahrenheit  [3] Kelvin")
            unit_options = {"1": "Celsius", "2": "Fahrenheit", "3": "Kelvin"}

            while True:
                val_input = input("Enter value to convert: ")
                try:
                    val_float = float(val_input)
                    break
                except ValueError:
                    print("Error: The temperature value must be a number. Try again.")

            while True:
                from_choice = input("Select unit to convert FROM (1, 2, 3 or name): ").strip().capitalize()
                from_u = unit_options.get(from_choice, from_choice)
                if from_u in unit_options.values():
                    break
                print(f"Error: Unknown unit '{from_choice}'. Must be 1, 2, 3 or Celsius, Fahrenheit, Kelvin.")

            while True:
                to_choice = input("Select unit to convert TO (1, 2, 3 or name): ").strip().capitalize()
                to_u = unit_options.get(to_choice, to_choice)
                
                try:
                    result = convert_temperature(val_float, from_u, to_u)
                    print(f"--> Result: {val_float} {from_u} = {result:.2f} {to_u}")
                    break
                except ValueError as e:
                    print(f"Error: {e}")

        elif answer_main == "2":
            print("\n--- Distance Converter ---")
            print("Available units: [1] Km  [2] Miles  [3] Meters")
            unit_options = {"1": "Km", "2": "Miles", "3": "Meters"}

            while True:
                val_input = input("Enter value to convert: ")
                try:
                    val_float = float(val_input)
                    if val_float < 0:
                        print("Error: Distance cannot be negative. Try again.")
                        continue
                    break
                except ValueError:
                    print("Error: The distance value must be a number. Try again.")

            while True:
                from_choice = input("Select unit to convert FROM (1, 2, 3 or name): ").strip().capitalize()
                from_u = unit_options.get(from_choice, from_choice)
                if from_u in unit_options.values():
                    break
                print(f"Error: Unknown unit '{from_choice}'. Must be 1, 2, 3 or Km, Miles, Meters.")

            while True:
                to_choice = input("Select unit to convert TO (1, 2, 3 or name): ").strip().capitalize()
                to_u = unit_options.get(to_choice, to_choice)
                
                try:
                    result = convert_distance(val_float, from_u, to_u)
                    print(f"--> Result: {val_float} {from_u} = {result:.2f} {to_u}")
                    break
                except ValueError as e:
                    print(f"Error: {e}")

        elif answer_main == "3":
            print("\n--- Weight Converter ---")
            print("Available units: [1] Kg  [2] Lbs  [3] Grams")
            unit_options = {"1": "Kg", "2": "Lbs", "3": "Grams"}

            while True:
                val_input = input("Enter value to convert: ")
                try:
                    val_float = float(val_input)
                    if val_float < 0:
                        print("Error: Weight cannot be negative. Try again.")
                        continue
                    break
                except ValueError:
                    print("Error: The weight value must be a number. Try again.")

            while True:
                from_choice = input("Select unit to convert FROM (1, 2, 3 or name): ").strip().capitalize()
                from_u = unit_options.get(from_choice, from_choice)
                if from_u in unit_options.values():
                    break
                print(f"Error: Unknown unit '{from_choice}'. Must be 1, 2, 3 or Kg, Lbs, Grams.")

            while True:
                to_choice = input("Select unit to convert TO (1, 2, 3 or name): ").strip().capitalize()
                to_u = unit_options.get(to_choice, to_choice)
                
                try:
                    result = convert_weight(val_float, from_u, to_u)
                    print(f"--> Result: {val_float} {from_u} = {result:.2f} {to_u}")
                    break
                except ValueError as e:
                    print(f"Error: {e}")

        else:
            print("Invalid option. Please choose 1, 2, 3, or Q.")


run_converter()
````
