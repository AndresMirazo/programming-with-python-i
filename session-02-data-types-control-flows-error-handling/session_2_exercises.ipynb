## Exercise 1: Messy Sensor Data Cleaner
**Objective:** Practice type conversion and error handling (`try/except`).

You are given a list of temperature readings from a faulty sensor. Some are integers, some are strings, and some are completely invalid data (like `None` or text). 

Write a function that loops through the list, attempts to convert each item to a float, and ignores invalid entries using a `try/except` block. Finally, return the average of the valid temperatures.

````python
def clean_and_average(raw_data: list):
    valid_temps = []
    for x in raw_data:
        try:
            float(x)
            valid_temps.append(float(x))
        except:
            pass
    sum = 0.00      
    for z in valid_temps:
        sum = sum + z
    avg = sum/len(valid_temps)

    return avg
sensor_readings = [22.5, "23.1", "error", None, 21, "22.0C", 19.8]
print(f"Average: {clean_and_average(sensor_readings)}")
````

## Exercise 2: Transit Kiosk Menu Simulator
**Objective:** Build a menu-driven application with nested conditionals and defensive input validation.

Create a `while` loop that displays a menu for a transit ticket kiosk. Ask the user for their destination zone (1, 2, or 3) and their age. Apply a 50% discount if they are under 18 or over 65. 

Use `try/except` to ensure they don't crash the program by typing letters instead of numbers. If they enter invalid input, print an error and restart the loop.

````python
def run_kiosk():
    while True:
        try:
            zone = int(input("Insert Zone"))
        except ValueError:
            print("Enter you zone again.")
            continue
        
        try:
            age = int(input("Insert Age"))
        except ValueError:
            print("Enter you age again.")
            continue
        
        price = 10

        available_zones = [1, 2, 3]
        if zone in available_zones:
            if zone == 2:
                price = price + 10
            elif zone == 3:
                price = price + 20

            if age < 18 or age > 65:
                price = price * 0.5
                print("50% discount assigned.")

            print(f"Ticket price: {price}€")
        else:
            print("Zone must be 1, 2 or 3.")
            continue

        again = input("Buy another ticket? Type (yes/no)")
        again = again.lower().strip()
        if again != "yes":
            print("Adios!")
            break

run_kiosk()
````

## Exercise 3: Three-Strike Quiz System
**Objective:** Use loops, counters, and conditionals to manage state.

Write a short quiz program. The user must answer a question correctly. They have exactly 3 attempts. 
- If they get it right, print a success message and `break` out of the loop.
- If they fail 3 times, print a failure message.
- **Defensive Code:** Ensure you handle edge cases like different capitalizations or extra spaces in their input (e.g., " Paris " should be accepted).

````python
def run_quiz():
    question = "What is the capital of France?"
    correct_answer = "paris"
    attempts = 3

    while True:

        if attempts > 0:
            answer = input(question)
            answer = answer.lower().strip()
            if answer == correct_answer:
                print("Correct Answer.")
                break
            else:
                attempts = attempts - 1
        else:
            print("No more attempts.")
            break

run_quiz()
````

## Exercise 5: Universal Unit Converter
**Objective:** Build a menu-driven unit converter with robust exception handling and clean function design.

Create a program that converts between different unit systems. The user selects a category from a menu, then inputs a value to convert.

**Requirements:**
- Support at least 3 categories: **Temperature** (°C ↔ °F ↔ K), **Distance** (km ↔ miles ↔ meters), and **Weight** (kg ↔ lbs ↔ grams).
- Each category must be handled by its own dedicated function.
- Use `try/except` to catch `ValueError` when the user enters non-numeric input.
- Use a `while` loop so the user can perform multiple conversions before choosing to quit.
- **Edge Case:** Reject physically impossible values (e.g., a temperature below absolute zero: -273.15°C).

````python
def convert_temperature(value, from_unit, to_unit):
    """Convert between Celsius, Fahrenheit, and Kelvin."""
    try:
        value + 0
    except TypeError:
        return "Enter correct value format."

    if from_unit == "celsius" and to_unit == "kelvin":
        result = value + 273.15
    elif from_unit == "celsius" and to_unit == "fahrenheit":
        result = value * 9 / 5 + 32
    elif from_unit == "fahrenheit" and to_unit == "celsius":
        result = (value - 32) * 5 / 9
    elif from_unit == "fahrenheit" and to_unit == "kelvin":
        result = (value - 32) * 5 / 9 + 273.15
    elif from_unit == "kelvin" and to_unit == "celsius":
        result = value - 273.15
    elif from_unit == "kelvin" and to_unit == "fahrenheit":
        result = (value - 273.15) * 9 / 5 + 32
    elif from_unit == to_unit:
        result = value
    else:
        return "Select correct unit."

    if to_unit == "kelvin" and result < 0:
        raise ValueError("Below absolute zero.")
    if to_unit == "celsius" and result < -273.15:
        raise ValueError("Below absolute zero.")
    if to_unit == "fahrenheit" and result < -459.67:
        raise ValueError("Below absolute zero.")

    return result
    

def convert_distance(value, from_unit, to_unit):
    """Convert between km, miles, and meters."""
    try:
        value + 0
    except TypeError:
        return "Enter correct value format."

    if value < 0:
        raise ValueError("Distance cant be negative!")

    if from_unit == "km" and to_unit == "miles":
        result = value * 0.6213
    elif from_unit == "km" and to_unit == "meters":
        result = value * 1000
    elif from_unit == "miles" and to_unit == "km":
        result = value * 1.6093
    elif from_unit == "miles" and to_unit == "meters":
        result = value * 1609.344
    elif from_unit == "meters" and to_unit == "km":
        result = value / 1000
    elif from_unit == "meters" and to_unit == "miles":
        result = value / 1609.344
    elif from_unit == to_unit:
        result = value
    else:
        return "Select correct unit."
    
    return result


def convert_weight(value, from_unit, to_unit):
    """Convert between kg, lbs, and grams."""
    try:
        value + 0
    except TypeError:
        return "Enter correct value format."

    if value < 0:
        raise ValueError("Weight cant be negative.")

    if from_unit == "kg" and to_unit == "lbs":
        result = value * 2.2046
    elif from_unit == "kg" and to_unit == "grams":
        result = value * 1000
    elif from_unit == "lbs" and to_unit == "kg":
        result = value * 0.4536
    elif from_unit == "lbs" and to_unit == "grams":
        result = value * 453.592
    elif from_unit == "grams" and to_unit == "kg":
        result = value / 1000
    elif from_unit == "grams" and to_unit == "lbs":
        result = value / 453.592
    elif from_unit == to_unit:
        result = value
    else:
        return "Select correct unit."

    return result

def run_converter():
    """Main loop: display menu, get user choice, call the right function."""
    while True:
        print("[1] Temperature [2] Distance [3] Weight [Q] Quit")
        choice = input("Choose converter option: ")
        choice = choice.lower().strip()

        if choice == "q":
            print("Adios!")
            break

        try:
            if choice == "1":
                value = float(input("Value: "))
                from_unit = input("From (celsius, fahrenheit, kelvin): ")
                from_unit = from_unit.lower().strip()
                to_unit = input("To (celsius, fahrenheit, kelvin): ")
                to_unit = to_unit.lower().strip()
                print(convert_temperature(value, from_unit, to_unit))
            elif choice == "2":
                value = float(input("Value: "))
                from_unit = input("From (km, miles, meters): ")
                from_unit = from_unit.lower().strip()
                to_unit = input("To (km, miles, meters): ")
                to_unit = to_unit.lower().strip()
                print(convert_distance(value, from_unit, to_unit))
            elif choice == "3":
                value = float(input("Value: "))
                from_unit = input("From (kg, lbs, grams): ")
                from_unit = from_unit.lower().strip()
                to_unit = input("To (kg, lbs, grams): ")
                to_unit = to_unit.lower().strip()
                print(convert_weight(value, from_unit, to_unit))
            else:
                print("Invalid option. Choose 1, 2, 3 or Q.")
                continue
        except ValueError:
            print("Invalid value. Try again.")

run_converter()
````

## Exercise 6: Password Strength Validator
**Objective:** Practice string methods, conditional logic, and loop-based analysis.

Write a function that evaluates the strength of a password and returns a score from 0 to 5. The user can keep testing passwords until they type `quit`.

**Scoring Rules (1 point each):**
1. Length is at least 8 characters.
2. Contains at least one uppercase letter.
3. Contains at least one lowercase letter.
4. Contains at least one digit.
5. Contains at least one special character (`!@#$%^&*`).

**Output:**
- 0–1: `Weak`
- 2–3: `Medium`
- 4–5: `Strong`

**Hint:** Use a `for` loop to iterate through each character of the password string and check its properties with `.isupper()`, `.isdigit()`, etc.

````python
def check_password_strength(password):
    """Evaluate a password and return a score (0-5) and a label."""
    score = 0
    special_chars = "!@#$%^&*"

    if len(password) >= 8:
        score = score + 1

    has_upper = False
    has_lower = False
    has_number = False
    has_special = False

    for x in range(len(password)):
        if password[x] in special_chars:
            has_special = True
        if password[x].isupper():
            has_upper = True
        if password[x].islower():
            has_lower = True
        if password[x].isdigit():
            has_number = True

    if has_upper:
        score = score + 1
    if has_lower:
        score = score + 1
    if has_number:
        score = score + 1
    if has_special:
        score = score + 1

    if score <= 1:
        strength_label = "Weak"
    elif score <= 3:
        strength_label = "Medium"
    else:
        strength_label = "Strong"

    return score, strength_label

def run_password_checker():
    """Loop: ask user for passwords until they type 'quit'."""
    while True:
        print("Password Tester")
        password = input("Enter a password (or type 'quit' to exit): ")

        check = password.lower().strip()
        if check == "quit":
            print("Adios!")
            break

        result = check_password_strength(password)
        score = result[0]
        label = result[1]
        print(f"Score: {score}/5 - {label}")

run_password_checker()
````

## Exercise 7: Multi-Round Scoring Quiz (with Data Structures)
**Objective:** Combine loops, conditionals, error handling, and dictionaries to build a stateful application.

Upgrade Exercise 3 into a full multi-question quiz. The questions and answers are stored in a **list of dictionaries**. The program tracks the user's score across all questions and displays a final report card.

**Requirements:**
- Store at least 5 questions in a list of dicts: `[{"question": ..., "answer": ..., "hint": ...}, ...]`
- The user gets **2 attempts per question**. After the first wrong answer, display the hint.
- Clean all user input (`.strip().lower()`).
- At the end, print a summary: total score, percentage, and a grade (A/B/C/F).
- Use `try/except` if any question expects a numeric answer.

````python
quiz_bank = [
    {"question": "What keyword creates a function in Python?", "answer": "def", "hint": "It's a 3-letter word."},
    {"question": "What data type is immutable: list or tuple?", "answer": "tuple", "hint": "It uses parentheses."},
    {"question": "What is 2 ** 10?", "answer": "1024", "hint": "It's a power of 2."},
    {"question": "What method removes whitespace from both ends of a string?", "answer": "strip", "hint": "Think .___()."},
    {"question": "What built-in function returns the number of items in a list?", "answer": "len", "hint": "Short for length."},
]

def run_full_quiz(questions):
    score = 0
    total = len(questions)

    for x in questions:
        attempts = 2

        while True:
            if attempts > 0:
                answer = input(x["question"] + " ")
                answer = answer.lower().strip()

                if x["answer"].isdigit():
                    try:
                        int(answer)
                    except ValueError:
                        print("Please enter a number.")

                if answer == x["answer"]:
                    print("Correct Answer.")
                    score = score + 1
                    break
                else:
                    attempts = attempts - 1
                    if attempts == 1:
                        print("Hint:", x["hint"])
            else:
                print("No more attempts. The answer was:", x["answer"])
                break

    percentage = score / total * 100

    if percentage >= 90:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 50:
        grade = "C"
    else:
        grade = "F"

    print(f"Total score: {score}/{total}")
    print(f"Percentage: {percentage}%")
    print(f"Grade: {grade}")

run_full_quiz(quiz_bank)
````

## Exercise 8: Expense Tracker with Category Reports
**Objective:** Build a complete menu-driven application that combines everything: loops, nested conditionals, error handling, functions, and data structures.

Create a mini expense tracker where the user can add expenses, view them by category, and see a summary report. This exercise ties together all the concepts from Session II.

**Menu Options:**
1. **Add Expense** — Ask for amount (float), category (string), and description (string). Validate that the amount is a positive number.
2. **View by Category** — Show all expenses for a user-selected category.
3. **Summary Report** — Print: total spent, number of transactions, average transaction, and the category with the highest spend.
4. **Quit**

**Data Structure:** Store expenses as a list of dictionaries:
```python
[{"amount": 12.50, "category": "food", "description": "Lunch"}, ...]
```

**Defensive Coding Requirements:**
- `try/except ValueError` for the amount input.
- Reject negative or zero amounts.
- Handle the case where the user asks for a summary with zero expenses (avoid `ZeroDivisionError`).

````python
expenses = []

def add_expense(expenses_list):
    """Prompt user for expense details and append to the list."""
    while True:
        try:
            amount = float(input("Amount: "))
        except ValueError:
            print("Enter the amount again.")
            continue

        if amount <= 0:
            print("Amount must be greater than 0.")
            continue
        break

    category = input("Category: ")
    category = category.lower().strip()
    description = input("Description: ")
    description = description.strip()

    expenses_list.append({"amount": amount, "category": category, "description": description})
    print("Expense added.")

def view_by_category(expenses_list):
    """Show expenses filtered by a user-selected category."""
    if len(expenses_list) == 0:
        print("No expenses yet.")
        return

    categories = []
    for x in expenses_list:
        if x["category"] not in categories:
            categories.append(x["category"])

    print("Categories:", categories)
    choice = input("Pick a category: ")
    choice = choice.lower().strip()

    found = False
    for x in expenses_list:
        if x["category"] == choice:
            description = x["description"]
            amount = x["amount"]
            print(f"{description}: {amount}€")
            found = True

    if found == False:
        print("No expenses in that category.")

def summary_report(expenses_list):
    """Print total, count, average, and top category."""
    if len(expenses_list) == 0:
        print("No expenses yet.")
        return

    total = 0.00
    for x in expenses_list:
        total = total + x["amount"]

    count = len(expenses_list)
    average = total / count

    categories = []
    for x in expenses_list:
        if x["category"] not in categories:
            categories.append(x["category"])

    top_category = ""
    top_amount = 0
    for c in categories:
        category_total = 0.00
        for x in expenses_list:
            if x["category"] == c:
                category_total = category_total + x["amount"]

        if category_total > top_amount:
            top_amount = category_total
            top_category = c

    print(f"Total spent: {total}€")
    print(f"Transactions: {count}")
    print(f"Average: {round(average, 2)}€")
    print(f"Top category: {top_category} ({top_amount}€)")

def run_tracker():
    """Main menu loop."""
    while True:
        print("[1] Add [2] View by Category [3] Summary [4] Quit")
        choice = input("Choose an option: ")
        choice = choice.strip()

        if choice == "1":
            add_expense(expenses)
        elif choice == "2":
            view_by_category(expenses)
        elif choice == "3":
            summary_report(expenses)
        elif choice == "4":
            print("Adios!")
            break
        else:
            print("Invalid option. Choose 1, 2, 3 or 4.")

run_tracker()
````
