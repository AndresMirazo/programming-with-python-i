## Exercise 2: Transit Kiosk Menu Simulator
**Objective:** Build a menu-driven application with nested conditionals and defensive input validation.

Create a `while` loop that displays a menu for a transit ticket kiosk. Ask the user for their destination zone (1, 2, or 3) and their age. Apply a 50% discount if they are under 18 or over 65. 

Use `try/except` to ensure they don't crash the program by typing letters instead of numbers. If they enter invalid input, print an error and restart the loop.

````python
def run_kiosk():
    while True:
        try:
            zone = int(input("Enter zone (1, 2, or 3): "))
            if zone not in (1, 2, 3):
                raise ValueError("Zone must be 1, 2, or 3.")
        except ValueError as error:
            print(error)
            continue

        try:
            age = int(input("Enter your age: "))
            if not 0 < age < 120:
                raise ValueError("Please enter an age between 1 and 119.")
        except ValueError as error:
            print(error)
            continue

        discount = 0.5 if age < 18 or age > 65 else 0
        print(f"You have selected zone {zone} and the discount is {discount * 100}%")
        break
run_kiosk()
````

## Exercise 5: Universal Unit Converter
**Objective:** Build a menu-driven unit converter with robust exception handling and clean function design.

Create a program that converts between different unit systems. The user selects a category from a menu, then inputs a value to convert.

**Requirements:**
- Support at least 3 categories: **Temperature** (°C ↔ °F ↔ K), **Distance** (km ↔ miles ↔ meters), and **Weight** (kg ↔ lbs ↔ grams).
- Each category must be handled by its own dedicated function.
- Use `try/except` to catch `ValueError` when the user enters non-numeric input.
- Use a `while` loop so the user can perform multiple conversions before choosing to quit.
- **Edge Case:** Reject physically impossible values (e.g., a temperature below absolute zero: -273.15°C).

````python
def convert_temperature(value, from_unit, to_unit):
    """Convert between Celsius, Fahrenheit, and Kelvin."""
    units = {"C", "F", "K"}
    if from_unit not in units or to_unit not in units:
        raise ValueError("Invalid temperature unit. Please use 'Celsius', 'Fahrenheit', or 'Kelvin'.")
    if from_unit == "C":
        celsius = value
    elif from_unit == "F":
        celsius = (value - 32) * 5 / 9
    else:
        celsius = value - 273.15
    if to_unit == "C":
        result = celsius
    elif to_unit == "F":
        result = (celsius * 9 / 5) + 32
    else:
        result = celsius + 273.15
    abs_zero = {
        "C": -273.15,
        "F": -459.67,
        "K": 0.0
    }
    # TODO: Implement conversion logic
    # TODO: Raise ValueError if result is below absolute zero
    if result < abs_zero[to_unit]:
        raise ValueError("Result is below absolute zero.")
    return result

def convert_distance(value, from_unit, to_unit):
    """Convert between km, miles, and meters."""
    if value < 0:
        raise ValueError("Distance cannot be negative.")

    from_unit = from_unit.lower()
    to_unit = to_unit.lower()

    if from_unit == "km":
        meters = value * 1000
    elif from_unit == "m":
        meters = value
    elif from_unit in ("mi", "mile", "miles"):
        meters = value * 1609.34
    else:
        raise ValueError("Unsupported distance unit.")

    if to_unit == "km":
        return meters / 1000
    elif to_unit == "m":
        return meters
    elif to_unit in ("mi", "mile", "miles"):
        return meters / 1609.34
    else:
        raise ValueError("Unsupported distance unit.")

def convert_weight(value, from_unit, to_unit):
    """Convert between kg, lbs, and grams."""
    if value < 0:
        raise ValueError("Weight cannot be negative.")

    from_unit = from_unit.lower()
    to_unit = to_unit.lower()

    if from_unit == "kg":
        kilograms = value
    elif from_unit in ("lb", "lbs", "pound", "pounds"):
        kilograms = value * 0.453592
    elif from_unit in ("g", "gram", "grams"):
        kilograms = value / 1000
    else:
        raise ValueError("Unsupported weight unit.")

    # Convert kilograms to target unit
    if to_unit == "kg":
        return kilograms
    elif to_unit in ("lb", "lbs", "pound", "pounds"):
        return kilograms / 0.453592
    elif to_unit in ("g", "gram", "grams"):
        return kilograms * 1000
    else:
        raise ValueError("Unsupported weight unit.")

def run_converter():
    """Main loop: display menu, get user choice, call the right function."""
    while True:
        print("\nMenu:")
        print("1. Temperature")
        print("2. Distance")
        print("3. Weight")
        print("Q. Quit")

        choice = input("Choose an option: ").strip().lower()

        if choice == "q":
            print("Goodbye!")
            break

        if choice not in ("1", "2", "3"):
            print("Invalid choice. Please enter 1, 2, 3, or Q.")
            continue

        try:
            value = float(input("Enter value: "))

            if choice == "1":
                from_unit = input("From unit (C, F, K): ").strip().upper()
                to_unit = input("To unit (C, F, K): ").strip().upper()
                result = convert_temperature(value, from_unit, to_unit)
                print(f"{value} {from_unit} = {result} {to_unit}")

            elif choice == "2":
                from_unit = input("From unit (km, m, mi): ").strip()
                to_unit = input("To unit (km, m, mi): ").strip()
                result = convert_distance(value, from_unit, to_unit)
                print(f"{value} {from_unit} = {result} {to_unit}")

            elif choice == "3":
                from_unit = input("From unit (kg, lb, g): ").strip()
                to_unit = input("To unit (kg, lb, g): ").strip()
                result = convert_weight(value, from_unit, to_unit)
                print(f"{value} {from_unit} = {result} {to_unit}")

        except ValueError as e:
            print(f"Error: {e}")
        except Exception:
            print("Invalid input. Please try again.")

run_converter()  # Uncomment to test
````
