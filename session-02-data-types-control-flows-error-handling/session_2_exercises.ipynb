## Exercise 1: Messy Sensor Data Cleaner
**Objective:** Practice type conversion and error handling (`try/except`).

You are given a list of temperature readings from a faulty sensor. Some are integers, some are strings, and some are completely invalid data (like `None` or text). 

Write a function that loops through the list, attempts to convert each item to a float, and ignores invalid entries using a `try/except` block. Finally, return the average of the valid temperatures.

````python
def clean_and_average(raw_data: list):
    valid_temps = []
    for elem in raw_data: 
        try:
            e = float(elem)
            valid_temps.append(e)
        except:
            pass
    avg = 0
    for elem in valid_temps:
        avg = avg + elem
    avg_div = avg / len(valid_temps)
    return avg_div

sensor_readings = [22.5, "23.1", "error", None, 21, "22.0C", 19.8]
print(f"Average: {clean_and_average(sensor_readings):.2f}")
````

## Exercise 2: Transit Kiosk Menu Simulator
**Objective:** Build a menu-driven application with nested conditionals and defensive input validation.

Create a `while` loop that displays a menu for a transit ticket kiosk. Ask the user for their destination zone (1, 2, or 3) and their age. Apply a 50% discount if they are under 18 or over 65. 

Use `try/except` to ensure they don't crash the program by typing letters instead of numbers. If they enter invalid input, print an error and restart the loop.

````python
def run_kiosk():
    while True:
        try:
            zone = int(input("What zone do you want to go (1, 2, or 3)? "))
            if zone not in [1, 2, 3]:
                print("Invalid zone. Please enter 1, 2, or 3.")
                continue
        except ValueError:
            print("Please enter the zone again.")
            continue
            
        try:
            age = int(input("How old are you? "))
        except ValueError:
            print("Please enter your age again.")
            continue        
            
        price = 10.0
        if zone == 2:
            price = 15.0
        elif zone == 3:
            price = 20.0
            
        if age < 18 or age > 65:
            price = price * 0.5
            
        print(f"Your ticket price is: ${price:.2f}")
            
        again = input("Do you want to buy another ticket (YES/NO)? ").strip().upper()
        if again == "NO":
            break

run_kiosk()
````

## Exercise 3: Three-Strike Quiz System
**Objective:** Use loops, counters, and conditionals to manage state.

Write a short quiz program. The user must answer a question correctly. They have exactly 3 attempts. 
- If they get it right, print a success message and `break` out of the loop.
- If they fail 3 times, print a failure message.
- **Defensive Code:** Ensure you handle edge cases like different capitalizations or extra spaces in their input (e.g., " Paris " should be accepted).

````python
def run_quiz():
    question = "What is the capital of France?"
    correct_answer = "paris"
    attempts = 3
    
    while True:
        answer = input(f"{question} Attempts: {attempts}: ").lower().strip()

        if answer == correct_answer:
            print("Correct!")
            break
        else:
            attempts -= 1
            print("Incorrect Please Try Again!")
        if attempts <= 0:
            print(f"No attemps left. Correct answer is {correct_answer}")
            break

run_quiz() # Uncomment to test
````

## Exercise 5: Universal Unit Converter
**Objective:** Build a menu-driven unit converter with robust exception handling and clean function design.

Create a program that converts between different unit systems. The user selects a category from a menu, then inputs a value to convert.

**Requirements:**
- Support at least 3 categories: **Temperature** (°C ↔ °F ↔ K), **Distance** (km ↔ miles ↔ meters), and **Weight** (kg ↔ lbs ↔ grams).
- Each category must be handled by its own dedicated function.
- Use `try/except` to catch `ValueError` when the user enters non-numeric input.
- Use a `while` loop so the user can perform multiple conversions before choosing to quit.
- **Edge Case:** Reject physically impossible values (e.g., a temperature below absolute zero: -273.15°C).

````python
def convert_temperature(value, from_unit, to_unit):
    """Convert between Celsius, Fahrenheit, and Kelvin."""
    # TODO: Implement conversion logic
    # TODO: Raise ValueError if result is below absolute zero
    from_unit = from_unit.upper()
    to_unit = to_unit.upper()

    if from_unit == 'F':
        celsius = (value - 32) * 5/9
    elif from_unit == 'K':
        celsius = value - 273.15
    elif from_unit == 'C':
        celsius = value
    else:
        raise ValueError("Invalid starting unit. Use C, F, or K.")

    if celsius < -273.15:
        raise ValueError("Temperature is below absolute zero.")

    if to_unit == 'F':
        return celsius * 9/5 + 32
    elif to_unit == 'K':
        return celsius + 273.15
    elif to_unit == 'C':
        return celsius
    else:
        raise ValueError("Invalid target unit. Use C, F, or K.")

def convert_distance(value, from_unit, to_unit):
    """Convert between km, miles, and meters."""
    # TODO: Implement conversion logic
    # TODO: Reject negative distances
    if value < 0:
        raise ValueError("Distance cannot be negative.")
        
    from_unit = from_unit.lower()
    to_unit = to_unit.lower()
    
    if from_unit == 'km':
        meters = value * 1000
    elif from_unit == 'miles':
        meters = value * 1609.34
    elif from_unit == 'meters':
        meters = value
    else:
        raise ValueError("Invalid starting unit. Use km, miles, or meters.")
        
    if to_unit == 'km':
        return meters / 1000
    elif to_unit == 'miles':
        return meters / 1609.34
    elif to_unit == 'meters':
        return meters
    else:
        raise ValueError("Invalid target unit. Use km, miles, or meters.")

def convert_weight(value, from_unit, to_unit):
    """Convert between kg, lbs, and grams."""
    # TODO: Implement conversion logic
    # TODO: Reject negative weights
    if value < 0:
        raise ValueError("Weight cannot be negative.")
        
    from_unit = from_unit.lower()
    to_unit = to_unit.lower()
    
    if from_unit == 'kg':
        grams = value * 1000
    elif from_unit == 'lbs':
        grams = value * 453.592
    elif from_unit == 'grams':
        grams = value
    else:
        raise ValueError("Invalid starting unit. Use kg, lbs, or grams.")
        
    if to_unit == 'kg':
        return grams / 1000
    elif to_unit == 'lbs':
        return grams / 453.592
    elif to_unit == 'grams':
        return grams
    else:
        raise ValueError("Invalid target unit. Use kg, lbs, or grams.")

def run_converter():
    """Main loop: display menu, get user choice, call the right function."""
    # TODO: while True loop with menu: [1] Temperature [2] Distance [3] Weight [Q] Quit
    # TODO: Inside each option, ask for value, from_unit, to_unit
    # TODO: Wrap everything in try/except to catch ValueError and bad input
    while True:
        print("\n*** Universal Unit Converter ***")
        print("[1] Temerature (C, F, K)")
        print("[2] Distance (km, miles, meters)")
        print("[3] Weight (kg, lbs, grams)")
        print("[Q] Quit")
        choice = input("Select an option: ").strip().upper()

        if choice == 'Q':
            break
        if choice in ['1', '2', '3']:
            try:
                value = float(input("Enter value to convert: "))
                from_unit = input("Enter starting unit: ").strip()
                to_unit = input("Enter target unit: ").strip()

                if choice == '1':
                    result = convert_temperature(value, from_unit, to_unit)
                elif choice == '2':
                    result = convert_distance(value, from_unit, to_unit)
                elif choice == '3':
                    result = convert_weight(value, from_unit, to_unit)

                print(f"Result: {result:.2f} {to_unit}")
            except ValueError as e:
                print(f"Error: {e}")
        else:
            print("Invalid menu choice. Please select 1, 2, 3, or Q.")

run_converter()  # Uncomment to test
````

## Exercise 6: Password Strength Validator
**Objective:** Practice string methods, conditional logic, and loop-based analysis.

Write a function that evaluates the strength of a password and returns a score from 0 to 5. The user can keep testing passwords until they type `quit`.

**Scoring Rules (1 point each):**
1. Length is at least 8 characters.
2. Contains at least one uppercase letter.
3. Contains at least one lowercase letter.
4. Contains at least one digit.
5. Contains at least one special character (`!@#$%^&*`).

**Output:**
- 0–1: `Weak`
- 2–3: `Medium`
- 4–5: `Strong`

**Hint:** Use a `for` loop to iterate through each character of the password string and check its properties with `.isupper()`, `.isdigit()`, etc.

````python
def check_password_strength(password):
    """Evaluate a password and return a score (0-5) and a label."""
    score = 0
    special_chars = "!@#$%^&*"
    
    # TODO: Check length >= 8
    if len(password) >= 8:
        score += 1
        print("password has length greater than 8!\n")
    # TODO: Loop through characters to check for uppercase, lowercase, digit, special
    upper_case_found, lower_case_found, digit_found, special_found = 0, 0, 0, 0
    for ch in password:
        if ch.isupper() and not upper_case_found:
            upper_case_found = 1
            score += 1
            print("Password has at least one upper case!\n")
        if ch.islower() and not lower_case_found:
            lower_case_found = 1
            score += 1
            print("password has at least one lower case!\n")
        if ch.isdigit() and not digit_found:
            digit_found = 1
            score += 1
            print("password has at least one digit!\n")
        if not special_found and ch in special_chars:
            special_found = 1
            score += 1
            print("password has at least one special character!\n")
    # TODO: Use boolean flags (has_upper, has_lower, etc.) and add to score
    # TODO: Return score and strength label
    return score

def run_password_checker():
    """Loop: ask user for passwords until they type 'quit'."""
    # TODO: while loop, get input, call check_password_strength, print result
    user_input = ""
    while user_input != 'quit':
        user_input = input("Enter Password to check strenght or 'quit' to exit: ")
        print(check_password_strength(user_input))

run_password_checker()  # Uncomment to test
````

## Exercise 7: Multi-Round Scoring Quiz (with Data Structures)
**Objective:** Combine loops, conditionals, error handling, and dictionaries to build a stateful application.

Upgrade Exercise 3 into a full multi-question quiz. The questions and answers are stored in a **list of dictionaries**. The program tracks the user's score across all questions and displays a final report card.

**Requirements:**
- Store at least 5 questions in a list of dicts: `[{"question": ..., "answer": ..., "hint": ...}, ...]`
- The user gets **2 attempts per question**. After the first wrong answer, display the hint.
- Clean all user input (`.strip().lower()`).
- At the end, print a summary: total score, percentage, and a grade (A/B/C/F).
- Use `try/except` if any question expects a numeric answer.

````python
quiz_bank = [
    {"question": "What keyword creates a function in Python?", "answer": "def", "hint": "It's a 3-letter word."},
    {"question": "What data type is immutable: list or tuple?", "answer": "tuple", "hint": "It uses parentheses."},
    {"question": "What is 2 ** 10?", "answer": "1024", "hint": "It's a power of 2."},
    {"question": "What method removes whitespace from both ends of a string?", "answer": "strip", "hint": "Think .___()."},
    {"question": "What built-in function returns the number of items in a list?", "answer": "len", "hint": "Short for length."},
]

def run_full_quiz(questions):
    score = 0
    total = len(questions)

    answers = []
    
    # TODO: Loop through each question dict
    # TODO: Give the user 2 attempts; show hint after the first failure
    # TODO: Clean user input before comparing
    # TODO: Track score
    for question in questions:
        correct = 0
        attempt = 1
        ans = ""
        while not correct and attempt < 3:
            if attempt == 2:
                print(f"Hint: {question['hint']}")
            ans = input(question['question']).strip().lower()

            if question["answer"].isdigit():
                try:
                    int(ans)
                except ValueError:
                    print("Invalid input! Please enter a number.")
            attempt += 1
            if ans == question["answer"]:
                score += 1
                break
        answers.append(ans)

    percentage = (score / total) * 100
    grade = ""
    if percentage >= 90:
        grade = 'A'
    elif percentage >= 70:
        grade = 'B'
    elif percentage >= 50:
        grade = 'C'
    else:
        grade = 'F'
    # TODO: Print final report card with score, percentage, and grade
    # Grading scale: >= 90% → A, >= 70% → B, >= 50% → C, else F
    print(f"Final report Card:\n\t Score: {score} \t Percentage: {percentage}% \t Grade: {grade}")
    for i in range(len(answers)):
        print(f"For question: {questions[i]["question"]}")
        print(f"You answered: {answers[i]} {"correctly" if answers[i] == questions[i]["answer"] else "wrongly correct answer was: ", questions[i]["answer"]}")
        

run_full_quiz(quiz_bank)  # Uncomment to test
````

## Exercise 8: Expense Tracker with Category Reports
**Objective:** Build a complete menu-driven application that combines everything: loops, nested conditionals, error handling, functions, and data structures.

Create a mini expense tracker where the user can add expenses, view them by category, and see a summary report. This exercise ties together all the concepts from Session II.

**Menu Options:**
1. **Add Expense** — Ask for amount (float), category (string), and description (string). Validate that the amount is a positive number.
2. **View by Category** — Show all expenses for a user-selected category.
3. **Summary Report** — Print: total spent, number of transactions, average transaction, and the category with the highest spend.
4. **Quit**

**Data Structure:** Store expenses as a list of dictionaries:
```python
[{"amount": 12.50, "category": "food", "description": "Lunch"}, ...]
```

**Defensive Coding Requirements:**
- `try/except ValueError` for the amount input.
- Reject negative or zero amounts.
- Handle the case where the user asks for a summary with zero expenses (avoid `ZeroDivisionError`).

````python
expenses = []

def add_expense(expenses_list):
    """Prompt user for expense details and append to the list."""
    # TODO: Ask for amount with try/except for ValueError
    # TODO: Reject negative or zero amounts
    amount, category, description = 0, "", ""
    while amount <= 0:
        try:
            amount = int(input("Enter amount greater than 0: "))
        except ValueError:
            print("Please enter a number!")
    
    # TODO: Ask for category and description
    category = input("Enter category: ").strip().lower()
    description = input("Enter description: ").strip()

    # TODO: Append a dict to expenses_list
    expenses_list.append({"amount": amount, "category": category, "description": description})

def view_by_category(expenses_list):
    """Show expenses filtered by a user-selected category."""
    # TODO: Get unique categories from the list
    # TODO: Ask user to pick one
    # TODO: Loop and print matching expenses
    if not expenses_list:
        print("No expenses recorded yet.")
        return

    categories = set(item["category"] for item in expenses_list)
    print("Available categories:", ", ".join(categories))
    
    chosen_category = input("Which category do you want to view? ").strip().lower()
    
    found = False
    for expense in expenses_list:
        if expense["category"] == chosen_category:
            print(f"${expense['amount']:.2f} - {expense['description']}")
            found = True
            
    if not found:
        print("No expenses found for that category.")

def summary_report(expenses_list):
    """Print total, count, average, and top category."""
    # TODO: Handle empty list case (no ZeroDivisionError!)
    # TODO: Calculate total, count, average
    # TODO: Find the category with the highest total spend
    if not expenses_list:
        print("No expenses to summarize.")
        return

    total = sum(expense["amount"] for expense in expenses_list)
    count = len(expenses_list)
    average = total / count
    
    category_totals = {}
    for expense in expenses_list:
        cat = expense["category"]
        category_totals[cat] = category_totals.get(cat, 0) + expense["amount"]
        
    top_category = max(category_totals, key=category_totals.get)
    
    print(f"Total spent: ${total:.2f}")
    print(f"Number of transactions: {count}")
    print(f"Average transaction: ${average:.2f}")
    print(f"Top spending category: {top_category} (${category_totals[top_category]:.2f})")

def run_tracker():
    """Main menu loop."""
    # TODO: while True with menu: [1] Add [2] View by Category [3] Summary [4] Quit
    # TODO: Use conditionals to route to the right function
    # TODO: Handle invalid menu choices gracefully
    while True:
        print("\n--- Expense Tracker ---")
        print("[1] Add Expense")
        print("[2] View by Category")
        print("[3] Summary Report")
        print("[4] Quit")
        
        choice = input("Select an option: ").strip()
        
        if choice == '1':
            add_expense(expenses)
        elif choice == '2':
            view_by_category(expenses)
        elif choice == '3':
            summary_report(expenses)
        elif choice == '4':
            break
        else:
            print("Invalid choice. Please enter 1, 2, 3, or 4.")

run_tracker()  # Uncomment to test
````
